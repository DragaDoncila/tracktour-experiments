```python
import pandas as pd
import os
from traccuracy.loaders import load_geff_data

GT_DIR = '/home/ddon0001/PhD/experiments/scaled/pre-thesis/scaled_w_merge/'
ds_names = [ds for ds in os.listdir(GT_DIR) if os.path.isdir(os.path.join(GT_DIR, ds))]
```

```python
ds_exp = []
skip_u = []
skip_v = []
skip_length = []
for ds_name in ds_names:
    gt_path = os.path.join(
        GT_DIR,
        ds_name,
        'matched_solution.zarr',
        'gt.geff'
    )
    gt_graph = load_geff_data(gt_path, load_all_props=True).graph
    for u, v in gt_graph.edges:
        u_t = gt_graph.nodes[u]['t']
        v_t = gt_graph.nodes[v]['t']
        if v_t - u_t > 1:
            ds_exp.append(ds_name)
            skip_u.append(u)
            skip_v.append(v)
            skip_length.append(v_t - u_t)
skip_df = pd.DataFrame({
    'ds_name': ds_exp,
    'skip_u': skip_u,
    'skip_v': skip_v,
    'skip_length': skip_length
})
```
