```python
import os

from tracktour_experiments.generate_dataset_summary import generate_ctc_summary

ROOT_DIR = '/home/ddon0001/PhD/data/cell_tracking_challenge/SUBMISSION/'
OUT_ROOT = '/home/ddon0001/PhD/experiments/scaled/pre-thesis/scaled_w_merge/'
out_csv_path = os.path.join(OUT_ROOT, 'summary.csv')

USE_ERR_SEG = True
INCLUDE_MERGES = True
USE_SCALE = True
USE_CTC_FORMAT = False
INCLUDE_BASIC_METRICS = True
```

```python
generate_ctc_summary(
    ROOT_DIR,
    out_csv_path,
    use_err_seg=USE_ERR_SEG
)
```

```python
from tracktour_experiments.generate_configs import ds_summary_to_configs
configs = ds_summary_to_configs(
    out_csv_path,
    OUT_ROOT,
    div_constraint=False,
    use_scale=USE_SCALE,
    allow_merges=INCLUDE_MERGES
)
```

```python
for config in configs:
     config.run(
          compute_additional_features=True,
          as_ctc=USE_CTC_FORMAT,          
     )
```

```python
for config in configs:
    print(config.data_config.dataset_name)
    config.evaluate(
        as_ctc=USE_CTC_FORMAT,
        include_basic_metrics=INCLUDE_BASIC_METRICS,
    )
```
